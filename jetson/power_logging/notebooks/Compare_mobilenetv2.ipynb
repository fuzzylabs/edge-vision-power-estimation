```python
import pandas as pd
import matplotlib.pyplot as plt

DATA_PATH = '../raw_data/mobilenet_v2_data'
avg_cols = ['average_power', 'average_run_time']
```

```python
trt_df = pd.read_csv(f'{DATA_PATH}/trt_mobilenet_v2_average_power_stats.csv')
```

```python
trt_df['layer_name'] = trt_df['layer_name'].apply(lambda x : x.split(' + ')[0].split('-')[-1].replace('[', '').replace(']', ''))
```

```python
subset_df = trt_df[['layer_name', 'average_power']]
subset_df.groupby(by=['layer_name'], sort=False).boxplot(subplots=False, figsize=(15, 15), rot=90, showfliers=False)
plt.show()
```

```python
subset_df = trt_df[['layer_name', 'average_run_time']]
subset_df.groupby(by=['layer_name'], sort=False).boxplot(subplots=False, figsize=(15, 15), rot=90, showfliers=False)
plt.show()
```

```python
subset_df = trt_df[['layer_name', 'average_power', 'average_run_time']]
subset_df['average_power'] = subset_df['average_power'] * 1e-3 # microwatts to millwats
subset_df['average_run_time'] = subset_df['average_run_time'] * 1e-3 # milliseconds to seconds
subset_df['average_energy'] = subset_df['average_power'] * subset_df['average_run_time']
subset_df = subset_df[['layer_name', 'average_energy']]
subset_df.groupby(by=['layer_name'], sort=False).boxplot(subplots=False, figsize=(25, 10), rot=90, showfliers=False)
plt.show()
```
